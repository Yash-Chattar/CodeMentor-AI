# CodeMentor AI — Model Experiments

This notebook reproduces the chronological model evaluation used by the Streamlit application.

**Important:** classification predicts the *next week's* performance category. The included CSV is synthetic demo data; do not interpret these metrics as real-world performance.

```python
import pandas as pd
from model_engine import validate_and_prepare, train_and_evaluate

DATA_PATH = "data/demo_coding_practice.csv"
raw = validate_and_prepare(pd.read_csv(DATA_PATH))
print(raw.shape)
raw.head()
```

```python
artifacts = train_and_evaluate(raw)

print("Selected classifier:", artifacts["best_classifier_name"])
print("Selected regressor:", artifacts["best_regressor_name"])
print("Final test begins at source week:", artifacts["test_start_week"])
```

```python
display(artifacts["classification_results"].round(3))
display(artifacts["regression_results"].round(3))
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

ConfusionMatrixDisplay(
    confusion_matrix=artifacts["confusion_matrix"],
    display_labels=["Weak", "Average", "Strong"]
).plot(cmap="Blues")
plt.title("Next-week performance category")
plt.show()
```

```python
from sklearn.metrics import mean_absolute_error, r2_score
import matplotlib.pyplot as plt

actual = artifacts["regression_test_actual"]
predicted = artifacts["regression_test_pred"]

plt.figure(figsize=(7, 6))
plt.scatter(actual, predicted, alpha=0.4)
plt.plot([0, 100], [0, 100], linestyle="--")
plt.xlabel("Actual next-week accuracy (%)")
plt.ylabel("Predicted next-week accuracy (%)")
plt.title("Actual vs predicted next-week accuracy")
plt.show()

print("MAE:", round(mean_absolute_error(actual, predicted), 3))
print("R²:", round(r2_score(actual, predicted), 3))
```
